1. Bank Account Manager

a. Simulate a bank account with deposit, withdrawal, and balance
enquiry operations

b. Use variables and data types to store account information

c. Implement conditional statements to check for sufficient balance
before withdrawal

d. Use functions to perform deposit, withdrawal, and balance inquiry
operations.

```python
class BankAccountManager:
  def __init__(self,owner,acc_no,balance):
    self.owner=owner
    self.acc_no=acc_no
    self.balance=balance
  def deposit(self,amount):
    if amount>=0:
      self.balance+=amount
      print(f"Deposited {amount:2f} Balance {self.balance:2f}")
    else:
      print("please deposit money")
  def withdraw(self,amount):
    if amount>=0:
      if self.balance>=amount:
        self.balance-=amount
        print(f"Withdraw {amount:2f} Balance {self.balance:2f}")
      else:
          print("insufficient balance")
    else:
      print("insufficient balance")
  def display(self):
    print(f"owner : {self.owner} acc_no:{self.acc_no:2f} balance :{self.balance:2f}")

cust=BankAccountManager("Abhi",18032348,10000)
cust.display()
cust.withdraw(20000)
```

Weather Data Analyzer

a. Analyze weather data (temperature, humidity, etc.) for a given region

b. Use variables and data types to store weather data

c. Implement conditional statements to determine weather conditions
(sunny, rainy, etc.)


d. Use functions to calculate average temperature and humidity, and print
results

```python
class WeatherDataAnalyzer:
  def __init__(self,temperature,humidity):
    self.temperature=temperature
    self.humidity=humidity
  def climate(self):
    average=(self.temperature+self.humidity)/2
    if average<25:
      print("the climate is rainy")
    elif average>=34:
      print("the climate is sunny")
    else:
      print("the climate is good")

weather=WeatherDataAnalyzer(35,20)
weather.climate()
```
